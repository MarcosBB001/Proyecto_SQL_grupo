```python
import db_utils 
```

```python
# Start the connection
conn, cur = db_utils.connect_to_db()
```

### Consulta 1

```python
conn.rollback()
```

```python
cur.execute("""SELECT a.nombre_alumno, p.nombre_promocion, ca.ciudad FROM alumno a
            INNER JOIN alumno_curso ac ON a.alumno_id=ac.alumno_id
            INNER JOIN curso c ON ac.curso_id=c.curso_id
            INNER JOIN promocion p ON c.promocion_id=p.promocion_id
            INNER JOIN campus ca ON c.campus_id=ca.campus_id
            WHERE p.nombre_promocion='Septiembre' AND ciudad='Madrid';
""")
alumnos_madrid_septiembre = cur.fetchall()
alumnos_madrid_septiembre
```

### Consulta 2

```python
cur.execute("""SELECT a.nombre_alumno FROM alumno a
        JOIN alumno_proyecto ap ON a.alumno_id = ap.alumno_id
        GROUP BY a.alumno_id, a.nombre_alumno
        HAVING COUNT(*) = SUM(CASE WHEN ap.apto = TRUE THEN 1 ELSE 0 END);
    """)
alumnos_aplicados = cur.fetchall()
alumnos_aplicados
```

### Consulta 3

```python
cur.execute("""SELECT bootcamp.nombre_bootcamp, COUNT(*) AS total_na FROM bootcamp
        JOIN curso ON bootcamp.bootcamp_id = curso.bootcamp_id
        JOIN alumno_proyecto ON curso.curso_id = alumno_proyecto.curso_id 
        WHERE alumno_proyecto.apto = false 
        GROUP BY bootcamp.nombre_bootcamp 
        ORDER BY total_na DESC LIMIT 2;
    """)
vertical_mayor_numero_no_aptos = cur.fetchall()
vertical_mayor_numero_no_aptos
```

### Consulta 4

```python
cur.execute("""SELECT p.nombre_profesor, b.nombre_bootcamp, ca.ciudad FROM profesor as p
        INNER JOIN prof_curso pc ON p.profesor_id=pc.profesor_id
        INNER JOIN curso c ON pc.curso_id=c.curso_id
        INNER JOIN bootcamp b ON c.bootcamp_id=b.bootcamp_id
        INNER JOIN campus ca ON c.campus_id=ca.campus_id
        WHERE b.nombre_bootcamp='DS' AND ca.ciudad='Madrid';
    """)
profesor_DS_madrid = cur.fetchall()
profesor_DS_madrid
```
